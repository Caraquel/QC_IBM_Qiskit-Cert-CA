# 2.1.4 Phase 2 Exam Drills

This notebook is a compact set of prediction drills for standard gates, phase behavior, and Bell-state outcomes.

## Instructions

Try to answer each prompt before executing the next code cell.

```python
import numpy as np
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector


def clean_probabilities(state, decimals=4):
    raw = state.probabilities_dict(decimals=decimals)
    return {str(k): float(v) for k, v in raw.items()}


def as_dict(state):
    return {str(k): complex(v) for k, v in state.to_dict().items()}
```

### Drill 1
Predict the final state of `X` followed by `T` on one qubit.

```python
qc = QuantumCircuit(1)
qc.x(0)
qc.t(0)
display(qc)
state = Statevector.from_instruction(qc)
print(as_dict(state))
```

### Drill 2
Predict the probabilities from `ry(pi/2)` on `|0>`.

```python
qc = QuantumCircuit(1)
qc.ry(np.pi / 2, 0)
display(qc)
state = Statevector.from_instruction(qc)
print(clean_probabilities(state))
```

### Drill 3
Predict whether `rz(pi/3)` changes computational-basis probabilities when applied directly to `|0>`.

```python
qc = QuantumCircuit(1)
qc.rz(np.pi / 3, 0)
display(qc)
state = Statevector.from_instruction(qc)
print("Statevector:", state.data)
print("Probabilities:", clean_probabilities(state))
```

### Drill 4
Predict the Bell-state probabilities from `H(0)` then `CX(0, 1)`.

```python
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
display(qc)
state = Statevector.from_instruction(qc)
print(clean_probabilities(state))
```

### Drill 5
Identify which states remain after this circuit.

```python
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc.z(0)
display(qc)
state = Statevector.from_instruction(qc)
print(as_dict(state))
```

## Final Challenge

Write your own circuit that produces `P(0) = 0.25` and `P(1) = 0.75` on a single qubit, then verify it below.

```python
# Replace this with your own solution.
qc = QuantumCircuit(1)
qc.ry(2 * np.arcsin(np.sqrt(0.75)), 0)
display(qc)
state = Statevector.from_instruction(qc)
print(clean_probabilities(state))
```
