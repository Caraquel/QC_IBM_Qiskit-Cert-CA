# 2.1.1 Phase Gates and the Bloch Sphere

This notebook focuses on the `Z`, `S`, and `T` gates and what they do to amplitudes, phase, and Bloch-sphere position.

## Core Idea

For a qubit in state `|1>`, the `S` gate multiplies the amplitude by `e^(i pi/2) = i` and the `T` gate multiplies it by `e^(i pi/4)`.

When the state is exactly `|1>`, this is a global phase. When the state is a superposition such as `|+>`, the phase becomes physically meaningful as a relative phase.

```python
import numpy as np
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector


def state_from_gate_sequence(gates, start_one=False):
    qc = QuantumCircuit(1)
    if start_one:
        qc.x(0)
    for gate in gates:
        getattr(qc, gate)(0)
    return qc, Statevector.from_instruction(qc)


def pretty_state(state):
    return {str(k): complex(v) for k, v in state.to_dict().items()}
```

## Z, S, and T Applied to `|1>`

```python
for gate in ["z", "s", "t"]:
    qc, state = state_from_gate_sequence([gate], start_one=True)
    display(qc)
    print(f"{gate.upper()} applied to |1> ->", pretty_state(state))
    print(f"Amplitude on |1>: {state.data[1]}")
    print("-" * 60)
```

```python
assert np.allclose(state_from_gate_sequence(["z"], start_one=True)[1].data[1], -1 + 0j)
assert np.allclose(state_from_gate_sequence(["s"], start_one=True)[1].data[1], 1j)
assert np.allclose(state_from_gate_sequence(["t"], start_one=True)[1].data[1], np.exp(1j * np.pi / 4))
print("Phase checks passed.")
```

## Relative Phase on `|+>`

This is where the phase gates become easier to see conceptually. Start with `|+> = (|0> + |1>) / sqrt(2)` and then apply `Z`, `S`, and `T`.

```python
for gate in ["z", "s", "t"]:
    qc = QuantumCircuit(1)
    qc.h(0)
    getattr(qc, gate)(0)
    state = Statevector.from_instruction(qc)
    display(qc)
    print(f"State after H then {gate.upper()}:", pretty_state(state))
    print("Bloch coordinates are easiest to inspect with the optional plot cell below.")
    print("-" * 60)
```

## Optional Bloch-Sphere Visualization

```python
try:
    from qiskit.visualization import plot_bloch_multivector

    qc = QuantumCircuit(1)
    qc.h(0)
    qc.t(0)
    state = Statevector.from_instruction(qc)
    plot_bloch_multivector(state, title="Bloch sphere after H then T")
except ImportError as exc:
    print("Visualization import skipped:", exc)
```
