# 2.1.3 Bell State and Standard Gates

This notebook practices the standard Bell-state pattern and connects it back to the single-qubit gates used to build it.

```python
import numpy as np
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector


def clean_probabilities(state, decimals=4):
    raw = state.probabilities_dict(decimals=decimals)
    return {str(k): float(v) for k, v in raw.items()}
```

## Build the Bell State

```python
bell = QuantumCircuit(2)
bell.h(0)
bell.cx(0, 1)
display(bell)

bell_state = Statevector.from_instruction(bell)
print("Amplitudes:", {str(k): complex(v) for k, v in bell_state.to_dict().items()})
print("Probabilities:", clean_probabilities(bell_state))
```

```python
probs = clean_probabilities(bell_state)
assert np.isclose(probs["00"], 0.5)
assert np.isclose(probs["11"], 0.5)
assert np.isclose(probs.get("01", 0.0), 0.0)
assert np.isclose(probs.get("10", 0.0), 0.0)
print("Bell-state probability checks passed.")
```

## Build Related Bell States

Add `X` or `Z` around the Bell-state sequence and predict which amplitudes change sign or basis index.

```python
variants = {}

phi_plus = QuantumCircuit(2)
phi_plus.h(0)
phi_plus.cx(0, 1)
variants["phi_plus"] = phi_plus

phi_minus = QuantumCircuit(2)
phi_minus.h(0)
phi_minus.z(0)
phi_minus.cx(0, 1)
variants["phi_minus"] = phi_minus

psi_plus = QuantumCircuit(2)
psi_plus.h(0)
psi_plus.cx(0, 1)
psi_plus.x(1)
variants["psi_plus"] = psi_plus

for name, circuit in variants.items():
    state = Statevector.from_instruction(circuit)
    print(name)
    display(circuit)
    print({str(k): complex(v) for k, v in state.to_dict().items()})
    print("-" * 60)
```
