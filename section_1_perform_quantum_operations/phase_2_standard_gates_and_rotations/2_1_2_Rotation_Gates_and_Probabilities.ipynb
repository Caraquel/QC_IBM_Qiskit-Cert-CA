# 2.1.2 Rotation Gates and Probabilities

This notebook practices `rx`, `ry`, and `rz` behavior, with a focus on how `ry` changes measurement probabilities.

## Core Idea

For certification questions, you should be able to predict amplitudes and measurement probabilities after basic single-qubit rotations.

```python
import numpy as np
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector


def clean_probabilities(state, decimals=4):
    raw = state.probabilities_dict(decimals=decimals)
    return {str(k): float(v) for k, v in raw.items()}


def run_rotation(gate_name, theta):
    qc = QuantumCircuit(1)
    getattr(qc, gate_name)(theta, 0)
    state = Statevector.from_instruction(qc)
    return qc, state
```

## Compare `rx`, `ry`, and `rz` at `theta = pi/2`

```python
for gate_name in ["rx", "ry", "rz"]:
    qc, state = run_rotation(gate_name, np.pi / 2)
    display(qc)
    print(f"{gate_name}(pi/2) statevector:", state.data)
    print(f"{gate_name}(pi/2) probabilities:", clean_probabilities(state))
    print("-" * 60)
```

## Key Exam Case: `ry(pi/2)` on `|0>`

```python
qc, state = run_rotation("ry", np.pi / 2)
print("Statevector:", state.data)
print("Probabilities:", clean_probabilities(state))
assert np.isclose(clean_probabilities(state)["0"], 0.5)
assert np.isclose(clean_probabilities(state)["1"], 0.5)
print("The 50/50 probability result is confirmed.")
```

## General Probability Pattern for `ry(theta)`

```python
angles = [0, np.pi / 6, np.pi / 4, np.pi / 2, np.pi]
for theta in angles:
    qc, state = run_rotation("ry", theta)
    probs = clean_probabilities(state)
    print(f"theta = {theta:.4f}")
    print("probabilities =", probs)
    print("expected P(1) = sin^2(theta/2) =", round(np.sin(theta / 2) ** 2, 4))
    print("-" * 60)
```

## Relative Phase Example With `rz`

`rz` changes phase relationships but not computational-basis measurement probabilities when applied directly to `|0>`.

```python
qc = QuantumCircuit(1)
qc.h(0)
qc.rz(np.pi / 2, 0)
state = Statevector.from_instruction(qc)
display(qc)
print("Statevector:", state.data)
print("Probabilities:", clean_probabilities(state))
```
