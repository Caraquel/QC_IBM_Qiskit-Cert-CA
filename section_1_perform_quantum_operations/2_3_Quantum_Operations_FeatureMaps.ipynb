# 2.3 Quantum Operations Feature Maps

This notebook practices data encoding with the current Qiskit 2.x function-based feature-map builders.

```python
import numpy as np
from qiskit.circuit.library import pauli_feature_map, z_feature_map, zz_feature_map
from qiskit.quantum_info import Statevector

FEATURES_2 = [0.2, 0.4]
FEATURES_3 = [0.2, 0.4, 0.8]
```

```python
def summarize_feature_map(name: str, circuit, values):
    encoded = circuit.assign_parameters(values)
    state = Statevector.from_instruction(encoded)
    print(name)
    display(encoded)
    print("Depth:", encoded.depth())
    print("Operation counts:", encoded.count_ops())
    print("Probabilities:", state.probabilities_dict(decimals=4))
    print("-" * 60)
```

## `z_feature_map`

```python
z_map = z_feature_map(feature_dimension=3, reps=1)
summarize_feature_map("z_feature_map", z_map, FEATURES_3)
```

## `zz_feature_map`

```python
zz_map = zz_feature_map(feature_dimension=3, reps=1, entanglement="linear")
summarize_feature_map("zz_feature_map", zz_map, FEATURES_3)
```

## `pauli_feature_map`

```python
pauli_map = pauli_feature_map(
    feature_dimension=2,
    reps=1,
    paulis=["Z", "ZZ", "XX"],
    entanglement="full",
)
summarize_feature_map("pauli_feature_map", pauli_map, FEATURES_2)
```

## Custom `data_map_func`

```python
def custom_phi(x):
    if len(x) == 1:
        return x[0]
    return (np.pi - x[0]) * (np.pi - x[1]) / 2

custom_zz = zz_feature_map(
    feature_dimension=2,
    reps=1,
    entanglement="full",
    data_map_func=custom_phi,
)
summarize_feature_map("zz_feature_map with custom data_map_func", custom_zz, FEATURES_2)
```
