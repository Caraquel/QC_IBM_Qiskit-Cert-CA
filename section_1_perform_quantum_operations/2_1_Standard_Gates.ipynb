# 2.1 Standard Gates

This notebook covers `Z`, `S`, `T`, rotation probabilities, Bell states, and optional state visualization.

```python
from pathlib import Path

import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

## Phase Gates On `|1>`

Exam detail: `S` adds a phase of `pi/2` to `|1>`, and `T` adds a phase of `pi/4` to `|1>`.

```python
def state_after_single_qubit_gate(gate_name: str) -> Statevector:
    qc = QuantumCircuit(1)
    qc.x(0)
    getattr(qc, gate_name)(0)
    return Statevector.from_instruction(qc)

expected = {
    "z": -1.0 + 0.0j,
    "s": 0.0 + 1.0j,
    "t": np.exp(1j * np.pi / 4),
}

for gate_name in ["z", "s", "t"]:
    state = state_after_single_qubit_gate(gate_name)
    amplitude = state.data[1]
    print(f"{gate_name.upper()}|1> amplitude on |1>: {amplitude}")
    assert np.allclose(amplitude, expected[gate_name])
```

## `ry(pi/2)` Probability Check

```python
qc = QuantumCircuit(1)
qc.ry(np.pi / 2, 0)
state = Statevector.from_instruction(qc)
probabilities = state.probabilities_dict(decimals=4)

print("Statevector:", state.data)
print("Measurement probabilities:", probabilities)
assert np.isclose(probabilities["0"], 0.5)
assert np.isclose(probabilities["1"], 0.5)
```

## Bell State

Create a Bell pair with `H` on qubit `0` followed by `CX(0, 1)`.

```python
bell = QuantumCircuit(2)
bell.h(0)
bell.cx(0, 1)

bell
```

```python
bell_state = Statevector.from_instruction(bell)
print("Amplitudes:", bell_state.to_dict(decimals=4))
print("Probabilities:", bell_state.probabilities_dict(decimals=4))
```

## Optional Visual Checks

```python
try:
    from qiskit.visualization import plot_histogram, plot_state_qsphere
    plot_histogram(bell_state.probabilities_dict(), title="Bell-State Measurement Probabilities")
except ImportError as exc:
    print("Visualization import skipped:", exc)
```

```python
try:
    from qiskit.visualization import plot_state_qsphere
    plot_state_qsphere(bell_state, show_state_phases=True)
except ImportError as exc:
    print("Visualization import skipped:", exc)
```
