# 2.3 Visualize Quantum States

This notebook focuses on Bloch-sphere and qsphere state visualization.

```python
import numpy as np
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
qc = QuantumCircuit(1)
qc.h(0)
qc.t(0)
display(qc)
state = Statevector.from_instruction(qc)
print(state.data)
```

## Bloch-Sphere View

```python
try:
    from qiskit.visualization import plot_bloch_multivector
    plot_bloch_multivector(state, title="Bloch sphere after H then T")
except ImportError as exc:
    print("Bloch-sphere visualization skipped:", exc)
```

## Qsphere View

```python
bell = QuantumCircuit(2)
bell.h(0)
bell.cx(0, 1)
bell_state = Statevector.from_instruction(bell)
try:
    from qiskit.visualization import plot_state_qsphere
    plot_state_qsphere(bell_state, show_state_phases=True)
except ImportError as exc:
    print("Qsphere visualization skipped:", exc)
```
