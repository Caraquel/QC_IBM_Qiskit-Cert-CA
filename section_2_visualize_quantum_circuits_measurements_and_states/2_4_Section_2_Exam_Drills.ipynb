# 2.4 Section 2 Exam Drills

Short drills for circuit, measurement, and state visualization.

```python
import numpy as np
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

### Drill 1
Predict the diagram and probabilities before running the next cell.

```python
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
display(qc)
state = Statevector.from_instruction(qc)
print({str(k): float(v) for k, v in state.probabilities_dict().items()})
```

### Drill 2
Does `rz(pi/2)` change computational-basis probabilities on `|0>`?

```python
qc = QuantumCircuit(1)
qc.rz(np.pi / 2, 0)
display(qc)
state = Statevector.from_instruction(qc)
print(state.data)
print({str(k): float(v) for k, v in state.probabilities_dict().items()})
```

### Drill 3
Inspect the Bloch-sphere view of `H` then `S`.

```python
qc = QuantumCircuit(1)
qc.h(0)
qc.s(0)
state = Statevector.from_instruction(qc)
try:
    from qiskit.visualization import plot_bloch_multivector
    plot_bloch_multivector(state, title="H then S")
except ImportError as exc:
    print("Bloch-sphere visualization skipped:", exc)
```

### Drill 4
Which amplitudes remain non-zero for this circuit?

```python
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc.z(0)
display(qc)
state = Statevector.from_instruction(qc)
print({str(k): complex(v) for k, v in state.to_dict().items()})
```
