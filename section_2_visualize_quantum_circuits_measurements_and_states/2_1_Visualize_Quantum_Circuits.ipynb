# 2.1 Visualize Quantum Circuits

This notebook introduces circuit visualization methods that are useful in Qiskit certification-style questions.

## References

This notebook follows the current Qiskit 2.x circuit drawing approach from the official IBM Quantum docs.

```python
from IPython.display import display
from qiskit import QuantumCircuit
```

```python
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])
qc
```

## Text Drawing

In Jupyter and VS Code, the last line of a cell often renders the circuit directly. You can also request a text drawer explicitly.

```python
print(qc.draw(output="text"))
```

## MPL Drawing

This depends on matplotlib support in your environment.

```python
try:
    qc.draw(output="mpl")
except Exception as exc:
    print("MPL drawer skipped:", exc)
```
