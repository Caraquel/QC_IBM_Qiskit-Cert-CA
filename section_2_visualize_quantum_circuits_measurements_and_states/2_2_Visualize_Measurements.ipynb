# 2.2 Visualize Measurements

This notebook focuses on measurement outcomes and histogram-based interpretation.

```python
import numpy as np
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
bell = QuantumCircuit(2)
bell.h(0)
bell.cx(0, 1)
display(bell)

state = Statevector.from_instruction(bell)
probs = {str(k): float(v) for k, v in state.probabilities_dict().items()}
print(probs)
```

## Histogram View

```python
try:
    from qiskit.visualization import plot_histogram
    plot_histogram(probs, title="Bell-state probabilities")
except ImportError as exc:
    print("Histogram skipped:", exc)
```

## Rotation Measurement Example

```python
qc = QuantumCircuit(1)
qc.ry(np.pi / 2, 0)
state = Statevector.from_instruction(qc)
probs = {str(k): float(v) for k, v in state.probabilities_dict().items()}
display(qc)
print(probs)
```

```python
assert np.isclose(probs["0"], 0.5)
assert np.isclose(probs["1"], 0.5)
print("Probability check passed.")
```
