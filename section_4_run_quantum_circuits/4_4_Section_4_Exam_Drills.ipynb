# 4.4 Section 4 Exam Drills

Short drills for transpilation and circuit execution.

```python
from IPython.display import display
from qiskit import QuantumCircuit, transpile
from qiskit.providers.fake_provider import GenericBackendV2
```

### Drill 1
Predict whether this circuit needs transpilation changes for the given coupling map.

```python
backend = GenericBackendV2(
    num_qubits=3,
    basis_gates=["id", "rz", "sx", "x", "cx", "measure"],
    coupling_map=[[0, 1], [1, 2]],
    seed=99,
)
qc = QuantumCircuit(3)
qc.cx(0, 2)
qc.measure_all()
display(qc)
transpiled = transpile(qc, backend=backend)
display(transpiled)
print(transpiled.count_ops())
```

### Drill 2
Predict the output counts of a deterministic circuit before running it.

```python
qc = QuantumCircuit(2)
qc.x(0)
qc.measure_all()
transpiled = transpile(qc, backend=backend)
counts = backend.run(transpiled, shots=256).result().get_counts()
print(counts)
```

### Drill 3
Run two circuits together and identify which result belongs to which circuit.

```python
qc1 = QuantumCircuit(1)
qc1.h(0)
qc1.measure_all()

qc2 = QuantumCircuit(1)
qc2.x(0)
qc2.measure_all()

transpiled_list = transpile([qc1, qc2], backend=GenericBackendV2(num_qubits=1, seed=5))
result = GenericBackendV2(num_qubits=1, seed=5).run(transpiled_list, shots=256).result()
print("Circuit 0:", result.get_counts(0))
print("Circuit 1:", result.get_counts(1))
```

### Drill 4
Compare two optimization levels.

```python
qc = QuantumCircuit(3)
qc.h(0)
qc.cx(0, 1)
qc.cx(1, 2)
qc.measure_all()
for level in [0, 3]:
    t_qc = transpile(qc, backend=backend, optimization_level=level)
    print(level, t_qc.depth(), t_qc.count_ops())
```
