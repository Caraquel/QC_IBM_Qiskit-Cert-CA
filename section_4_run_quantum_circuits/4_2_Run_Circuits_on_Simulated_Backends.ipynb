# 4.2 Run Circuits on Simulated Backends

This notebook focuses on executing transpiled circuits with a backend interface.

```python
from IPython.display import display
from qiskit import QuantumCircuit, transpile
from qiskit.providers.fake_provider import GenericBackendV2
```

## Prepare a Circuit and Backend

```python
backend = GenericBackendV2(num_qubits=2, seed=2026)
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()
display(qc)
transpiled = transpile(qc, backend)
display(transpiled)
```

## Run the Circuit

```python
job = backend.run(transpiled, shots=1024)
result = job.result()
counts = result.get_counts()
print("Job ID:", job.job_id())
print("Counts:", counts)
```

## Compare Different Shot Counts

```python
for shots in [100, 1000, 4000]:
    job = backend.run(transpiled, shots=shots)
    counts = job.result().get_counts()
    print(f"shots={shots} -> {counts}")
```

## Run More Than One Circuit

```python
qc2 = QuantumCircuit(2)
qc2.x(1)
qc2.measure_all()
transpiled_list = transpile([qc, qc2], backend)
job = backend.run(transpiled_list, shots=512)
result = job.result()
print("Circuit 0 counts:", result.get_counts(0))
print("Circuit 1 counts:", result.get_counts(1))
```
