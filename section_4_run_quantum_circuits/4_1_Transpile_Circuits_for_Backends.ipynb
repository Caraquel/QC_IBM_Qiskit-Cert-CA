# 4.1 Transpile Circuits for Backends

This notebook introduces backend-aware transpilation for execution.

## Core Idea

Abstract circuits are convenient for programming, but hardware-facing execution requires circuits that obey a backend target, including supported basis gates and connectivity.

```python
from IPython.display import display
from qiskit import QuantumCircuit, transpile
from qiskit.providers.fake_provider import GenericBackendV2
```

## Create an Abstract Circuit

```python
qc = QuantumCircuit(3)
qc.h(0)
qc.cx(0, 2)
qc.cz(1, 2)
qc.measure_all()
display(qc)
print("Original depth:", qc.depth())
print("Original ops:", qc.count_ops())
```

## Build a Simulated Backend Target

```python
backend = GenericBackendV2(
    num_qubits=3,
    basis_gates=["id", "rz", "sx", "x", "cx", "measure"],
    coupling_map=[[0, 1], [1, 2]],
    seed=1234,
)
print(backend)
```

## Transpile for That Backend

```python
transpiled = transpile(qc, backend=backend, optimization_level=1)
display(transpiled)
print("Transpiled depth:", transpiled.depth())
print("Transpiled ops:", transpiled.count_ops())
```

## Compare Optimization Levels

```python
for level in [0, 1, 2, 3]:
    t_qc = transpile(qc, backend=backend, optimization_level=level)
    print(f"optimization_level={level}")
    print("depth =", t_qc.depth())
    print("ops =", t_qc.count_ops())
    print("-" * 60)
```
