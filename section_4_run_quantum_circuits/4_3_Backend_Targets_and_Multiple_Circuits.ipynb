# 4.3 Backend Targets and Multiple Circuits

This notebook focuses on backend details that affect execution and on running lists of circuits.

```python
from IPython.display import display
from qiskit import QuantumCircuit, transpile
from qiskit.providers.fake_provider import GenericBackendV2
```

## Inspect Backend Capabilities

```python
backend = GenericBackendV2(
    num_qubits=3,
    basis_gates=["id", "rz", "sx", "x", "cx", "measure"],
    coupling_map=[[0, 1], [1, 2]],
    seed=77,
)
print("num_qubits:", backend.num_qubits)
print("operation_names:", backend.operation_names)
print("coupling_map:", backend.coupling_map)
```

## Create Several Circuits

```python
circuits = []

qc_a = QuantumCircuit(3)
qc_a.h(0)
qc_a.cx(0, 1)
qc_a.measure_all()
circuits.append(qc_a)

qc_b = QuantumCircuit(3)
qc_b.x(2)
qc_b.measure_all()
circuits.append(qc_b)

qc_c = QuantumCircuit(3)
qc_c.h(1)
qc_c.cx(1, 2)
qc_c.measure_all()
circuits.append(qc_c)

for i, circuit in enumerate(circuits):
    print(f"Circuit {i}")
    display(circuit)
```

## Transpile and Run the List

```python
transpiled_list = transpile(circuits, backend=backend, optimization_level=1)
job = backend.run(transpiled_list, shots=512)
result = job.result()
for i in range(len(transpiled_list)):
    print(f"Circuit {i} transpiled ops:", transpiled_list[i].count_ops())
    print(f"Circuit {i} counts:", result.get_counts(i))
    print("-" * 60)
```

## Compare a Custom Basis-Gate Target

```python
backend2 = GenericBackendV2(
    num_qubits=3,
    basis_gates=["id", "u", "cx", "measure"],
    coupling_map=[[0, 1], [1, 2]],
    seed=77,
)
alt = transpile(qc_a, backend=backend2, optimization_level=1)
display(alt)
print(alt.count_ops())
```
