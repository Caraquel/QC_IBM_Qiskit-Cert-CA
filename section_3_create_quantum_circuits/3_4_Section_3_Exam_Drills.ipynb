# 3.4 Section 3 Exam Drills

Short construction drills for Qiskit circuit-creation syntax and behavior.

```python
import numpy as np
from IPython.display import display
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit.circuit import Parameter
from qiskit.quantum_info import Statevector
```

### Drill 1
Predict the number of qubits, clbits, and instructions before running the next cell.

```python
qc = QuantumCircuit(2, 1)
qc.h(0)
qc.cx(0, 1)
display(qc)
print(qc.num_qubits, qc.num_clbits, qc.size())
```

### Drill 2
Predict where the measurement result is stored.

```python
qc = QuantumCircuit(2, 2)
qc.x(1)
qc.measure(1, 0)
display(qc)
print(qc.data)
```

### Drill 3
Predict the remaining parameters after assignment.

```python
theta = Parameter("theta")
phi = Parameter("phi")
qc = QuantumCircuit(1)
qc.ry(theta, 0)
qc.rz(phi, 0)
bound = qc.assign_parameters({theta: np.pi / 2})
display(bound)
print(bound.parameters)
```

### Drill 4
Predict the output state after a circuit and its inverse.

```python
qc = QuantumCircuit(1)
qc.h(0)
qc.t(0)
full = QuantumCircuit(1)
full.compose(qc, inplace=True)
full.compose(qc.inverse(), inplace=True)
display(full)
state = Statevector.from_instruction(full)
print(state.to_dict())
```

### Drill 5
Build a register-based circuit and inspect its resources.

```python
q = QuantumRegister(2, "q")
c = ClassicalRegister(2, "c")
qc = QuantumCircuit(q, c)
qc.h(q[0])
qc.measure(q, c)
display(qc)
print(qc.qregs, qc.cregs, qc.width())
```
