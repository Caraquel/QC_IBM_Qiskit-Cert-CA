# 3.1 Construct Basic Circuits

This notebook introduces the basic `QuantumCircuit` workflow for creating circuits in Qiskit.

## Core Idea

A quantum circuit is a collection of qubits plus a list of instructions that act on them. For certification questions, you should be comfortable building circuits from scratch and predicting what the code creates.

```python
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

## Create a Circuit With Integer Sizes

```python
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])
display(qc)
print("num_qubits:", qc.num_qubits)
print("num_clbits:", qc.num_clbits)
print("size:", qc.size())
print("depth:", qc.depth())
```

## Inspect Circuit Data

Each appended instruction appears in `qc.data`.

```python
ideal = QuantumCircuit(2)
ideal.h(0)
ideal.cx(0, 1)
display(ideal)
print("Circuit data entries:")
for item in ideal.data:
    print(item)
```

## Exact State Check

```python
state = Statevector.from_instruction(ideal)
print("Statevector amplitudes:", state.to_dict())
print("Probabilities:", state.probabilities_dict())
```
