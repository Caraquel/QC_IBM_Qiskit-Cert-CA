# 3.2 Registers, Bits, and Measurements

This notebook focuses on explicit register-based circuit construction.

```python
from IPython.display import display
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, AncillaRegister
from qiskit.quantum_info import Statevector
```

## Build a Circuit With Named Registers

```python
data = QuantumRegister(2, "data")
anc = AncillaRegister(1, "anc")
readout = ClassicalRegister(3, "c")

qc = QuantumCircuit(data, anc, readout, name="register_demo")
qc.h(data[0])
qc.cx(data[0], data[1])
qc.x(anc[0])
qc.measure(data[0], readout[0])
qc.measure(data[1], readout[1])
qc.measure(anc[0], readout[2])
display(qc)
```

## Inspect Circuit Structure

```python
print("qubits:", qc.qubits)
print("clbits:", qc.clbits)
print("qregs:", qc.qregs)
print("cregs:", qc.cregs)
print("num_ancillas:", qc.num_ancillas)
print("width:", qc.width())
```

## Selective Measurement Mapping

You should be comfortable measuring qubits into chosen classical bits, not only in matching order.

```python
qc2 = QuantumCircuit(2, 2)
qc2.x(1)
qc2.measure(1, 0)
qc2.measure(0, 1)
display(qc2)
print(qc2.data)
```

## Pre-measurement State Check

```python
ideal = QuantumCircuit(data, anc)
ideal.h(data[0])
ideal.cx(data[0], data[1])
ideal.x(anc[0])
state = Statevector.from_instruction(ideal)
print(state.to_dict())
```
