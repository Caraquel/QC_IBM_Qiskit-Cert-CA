# 3.3 Parameters, Compose, and Inverse

This notebook covers parameterized circuits and subcircuit composition.

```python
import numpy as np
from IPython.display import display
from qiskit import QuantumCircuit
from qiskit.circuit import Parameter, ParameterVector
from qiskit.quantum_info import Statevector
```

## Single Parameters

```python
theta = Parameter("theta")
qc = QuantumCircuit(1)
qc.ry(theta, 0)
display(qc)
print("Parameters:", qc.parameters)
```

```python
bound = qc.assign_parameters({theta: np.pi / 2})
display(bound)
state = Statevector.from_instruction(bound)
print(state.probabilities_dict())
```

## ParameterVector

```python
angles = ParameterVector("a", 3)
layer = QuantumCircuit(3, name="layer")
for i, angle in enumerate(angles):
    layer.ry(angle, i)
layer.cx(0, 1)
layer.cx(1, 2)
display(layer)
print(layer.parameters)
```

## Compose a Subcircuit

```python
qc2 = QuantumCircuit(3)
qc2.h([0, 1, 2])
qc2.compose(layer, inplace=True)
bound2 = qc2.assign_parameters([np.pi / 6, np.pi / 4, np.pi / 3])
display(bound2)
print("depth:", bound2.depth())
```

## Inverse Check

```python
undo = QuantumCircuit(3)
undo.compose(layer, inplace=True)
undo.compose(layer.inverse(), inplace=True)
bound_undo = undo.assign_parameters([0.2, 0.4, 0.8])
display(bound_undo)
state = Statevector.from_instruction(bound_undo)
print(state.to_dict())
```
